# Data Analysis Demo

This notebook demonstrates interleaved text and chart outputs.

```
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)
print(f'Mean: {data.mean():.4f}')
print(f'Std:  {data.std():.4f}')
```

```
plt.figure(figsize=(8, 4))
plt.hist(data, bins=30, color='steelblue', edgecolor='white')
plt.title('Distribution of Random Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Scatter Plot

Let's look at correlations between two variables.

```
x = np.linspace(0, 10, 50)
y = 2 * x + np.random.randn(50) * 2
plt.scatter(x, y, alpha=0.7)
plt.plot(x, 2*x, 'r--', label='y=2x')
plt.legend()
plt.title('Scatter with Trend')
plt.show()
```

```
summary = {'count': 50, 'r_squared': 0.92}
print(summary)
plt.bar(['A', 'B', 'C'], [3, 7, 5])
plt.title('Category Counts')
plt.show()
```
